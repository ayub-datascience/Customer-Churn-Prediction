# Customer Churn Prediction

## Define the Business Problem

### 1.1 Business Problem

The objective of this project is to predict whether an active customer
will churn based on historical customer behavior, subscription
information, payment behavior, spending, and engagement characteristics.

The prediction can help a business identify customers who may be at
higher risk of leaving and support customer-retention activities.

### 1.2 What is Customer Churn?

**Churn** refers to a customer discontinuing or cancelling their
subscription or service.

For this project:

- `Churn = 1` → Customer churned
- `Churn = 0` → Customer did not churn

### 1.3 Prediction Target

The target variable is:

**`Churn`**

This is the variable that our machine learning model will learn to predict.

### 1.4 Machine Learning Problem

This project is a:

- **Learning Type:** Supervised Learning
- **Problem Type:** Binary Classification
- **Target Variable:** `Churn`

The model will learn from historical customer records and predict
whether a customer belongs to the churn or non-churn class.

### 1.5 Input Features

The model will use relevant customer characteristics such as:

- Customer age
- Customer tenure
- Service usage
- Number of support calls
- Payment delays
- Subscription type
- Contract length
- Total spending
- Last interaction 

### 1.6 Business Objective

The main business objective is to identify customers who are at risk
of churn so that the business can potentially take appropriate
customer-retention actions.

### 1.7 Success Criteria

The model should:

1. Identify customers who are likely to churn.
2. Perform well on unseen customer data.
3. Provide reliable classification results.
