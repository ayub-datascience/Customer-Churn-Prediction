# What characteristics are associated with customers leaving?

## We can analyze it by plotting the graphs.

```python
import pandas as pd
```

```python
df = pd.read_csv("cleaned_customer_churn_data.csv")
```

```python
df
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.lineplot(data=df, x="Churn", y="Tenure")
plt.xlabel('Churn')
plt.ylabel('Tenure')
plt.show()
```

```python
sns.lineplot(data=df, x="Churn", y="Usage Frequency")
plt.xlabel('Churn')
plt.ylabel('Usage Frequency')
plt.show()
```

```python
sns.lineplot(data=df, x="Churn", y="Support Calls")
plt.xlabel('Churn')
plt.ylabel('Support Calls')
plt.show()
```

```python
sns.lineplot(data=df, x="Churn", y="Payment Delay")
plt.xlabel('Churn')
plt.ylabel('Payment Delay')
plt.show()
```

```python
sns.lineplot(data=df, x="Churn", y="Total Spend")
plt.xlabel('Churn')
plt.ylabel('Total Spend')
plt.show()
```

```python
sns.lineplot(data=df, x="Churn", y="Last Interaction")
plt.xlabel('Churn')
plt.ylabel('Last Interaction')
plt.show()
```

```python
sns.lineplot(data=df, x="Churn", y="Age")
plt.xlabel('Churn')
plt.ylabel('Age')
plt.show()
```

```python
sns.countplot(data=df, x="Churn")
plt.show()
```

```python
sns.heatmap(df.corr(numeric_only=True), annot=True)
plt.show()
```
