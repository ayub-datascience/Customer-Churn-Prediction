# Splitting the data into train and test sets.

## Before moving ahead to preprocessing stage, We separate the data into train and test sets mainly to prevent data leakage.

### Here we are splitting the data using the reference of 'Churn'. Because we need to ensure that each and every value of 'Churn' is equally distributed into train and test sets. Hence we use 'StratifiedShuffleSplit' method.

```python
import pandas as pd
```

```python
df = pd.read_csv("cleaned_customer_churn_data.csv")
```

```python
df
```

```python
# StratifiedShuffleSplit
from sklearn.model_selection import StratifiedShuffleSplit
```

```python
split = StratifiedShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=42
)
```

```python
for train_index, test_index in split.split(df, df['Churn']):
    strat_train_set = df.iloc[train_index]
    strat_test_set = df.iloc[test_index]
```

```python
strat_train_set
```

```python
strat_test_set
```

```python
# Saving the training and testing data into two different csv files.
strat_train_set.to_csv("training_data.csv", index=False)
strat_test_set.to_csv("testing_data.csv", index=False)
print("Files saved successfully")
```
