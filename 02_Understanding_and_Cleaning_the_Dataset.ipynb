```python
# In the dataset, Every column contains one missing value i.e NaN
print(df.isnull().sum())
```

```python
# In the dataset, Every column contains one missing value at the index "199295". Now the better idea is to remove the entire row.
for column in df.columns:
    missing_indices = df.index[df[column].isnull()].tolist()
    
    if missing_indices:
        print(f"{column} -> Row index: {missing_indices}")
```

```python
# Remove rows where all values are missing
df = df.dropna(how="all")

print("New shape:", df.shape)
```

```python
# Verifying that the row was removed
print("Missing values after removal:")
print(df.isnull().sum())
```

```python
print("Number of duplicate records:", df.duplicated().sum())
```

```python
# Categorical variables
categorical_columns = df.select_dtypes(include=["object", "category"]).columns

print("Categorical variables:")
print(categorical_columns.tolist())
```

```python
# Numerical variables
numerical_columns = df.select_dtypes(include=["int64", "float64"]).columns

print("Numerical variables:")
print(numerical_columns.tolist())
```

```python
# Target (Label) distribution
print(df["Churn"].value_counts())
```

```python
# There is a column called "CustomerID" in the dataset, which is not useful. Hence removing or dropping the entire column.
df = df.drop(columns=["CustomerID"])
```

```python
# Verifying that "CustomerID" is removed successfully
print(df.columns)
```

```python
# Saving the cleaned DataFrame as a new CSV
df.to_csv("cleaned_customer_churn_data.csv", index=False)
```
