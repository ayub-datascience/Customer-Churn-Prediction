```python
# Cross validation method is used to avoid overfitting and To choose best model.
```

```python
# I want to identify as many customers likely to churn as possible; So I would use "recall" as the primary cross-validation scoring metric.
```

```python
import pandas as pd

X_train = pd.read_csv("training_data_preprocessed.csv")
y_train = pd.read_csv("training_data.csv")["Churn"]
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.model_selection import cross_val_score
```

```python
logistic_model = LogisticRegression(
    random_state=42
)
```

```python
decision_tree_model = DecisionTreeClassifier(
    random_state=42
)
```

```python
random_forest_model = RandomForestClassifier(
    random_state=42
)
```

```python
gradient_boosting_model = GradientBoostingClassifier(
    random_state=42
)
```

```python
# 1. Evaluate Logistic Regression with cross-validation

logistic_recalls = cross_val_score(
    logistic_model, # The model we want to evaluate
    X_train, # Features
    y_train, # Labels
    scoring="recall",
    cv=10 # Number of folds for cross-validation (10-fold cross-validation means the data is split into 10 parts, and the model is trained and evaluated 10 times, each time using a different part as the test set and the remaining parts as the training set)
)

print("Logistic Regression cross validation recall's: ", logistic_recalls)
```

```python
# 2. Evaluate Decision Tree Classifier with cross-validation

decision_tree_recalls = cross_val_score(
    decision_tree_model, # The model we want to evaluate
    X_train, # Features
    y_train, # Labels
    scoring="recall",
    cv=10 # Number of folds for cross-validation (10-fold cross-validation means the data is split into 10 parts, and the model is trained and evaluated 10 times, each time using a different part as the test set and the remaining parts as the training set)
)

print("Decision Tree Classifier cross validation recall's: ", decision_tree_recalls)
```

```python
# 3. Evaluate Random Forest Classifier with cross-validation

random_forest_recalls = cross_val_score(
    random_forest_model, # The model we want to evaluate
    X_train, # Features
    y_train, # Labels
    scoring="recall",
    cv=10 # Number of folds for cross-validation (10-fold cross-validation means the data is split into 10 parts, and the model is trained and evaluated 10 times, each time using a different part as the test set and the remaining parts as the training set)
)

print("Random Forest Classifier cross validation recall's: ", random_forest_recalls)
```

```python
# 4. Evaluate Gradient Boosting Classifier with cross-validation

gradient_boosting_recalls = cross_val_score(
    gradient_boosting_model, # The model we want to evaluate
    X_train, # Features
    y_train, # Labels
    scoring="recall",
    cv=10 # Number of folds for cross-validation (10-fold cross-validation means the data is split into 10 parts, and the model is trained and evaluated 10 times, each time using a different part as the test set and the remaining parts as the training set)
)

print("Gradient Boosting Classifier cross validation recall's: ", gradient_boosting_recalls)
```

# According to these cross-validation results on each model "Decision Tree Classifier" is the BEST FIT

## The reason behind is:

### The Decision Tree achieved the highest mean 10-fold cross-validation Recall (~99.991%), with extremely consistent performance across all folds. Since the primary objective is to identify as many customers likely to churn as possible, minimizing false negatives is the priority, making Recall the most important selection metric.
