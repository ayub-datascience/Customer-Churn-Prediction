```python
import pandas as pd
```

```python
df = pd.read_csv("training_data.csv")
```

```python
df
```

```python
# Seperate Features and Label
x_train = df.drop("Churn", axis=1)
y_train = df["Churn"]
```

```python
x_train # Features
```

```python
y_train # Label
```

```python
# Seperate Numerical and Categorical column names in the form of python list.
num_attribs = ["Age", "Tenure", "Usage Frequency", "Support Calls", "Payment Delay", "Total Spend", "Last Interaction"]
cat_attribs = ["Gender", "Subscription Type", "Contract Length"]
```

```python
num_attribs
```

```python
cat_attribs
```

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
```

```python
# Numerical Pipeline
num_pipeline = Pipeline([
    ("scaler", StandardScaler())
])
```

```python
# Categorical Pipeline
cat_pipeline = Pipeline([
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])
```

```python
# Full Pipeline
full_pipeline = ColumnTransformer([
    ("num", num_pipeline, num_attribs),
    ("cat", cat_pipeline, cat_attribs)
])
```

```python
# Fit and Transform the data
customer_churn_prepared = full_pipeline.fit_transform(x_train)
```

```python
customer_churn_prepared
```

```python
# Converting it back to pandas Dataframe
feature_names = full_pipeline.get_feature_names_out()

preprocessed = pd.DataFrame(
    customer_churn_prepared,
    columns=feature_names,
    index=x_train.index
)
```

```python
preprocessed
```

```python
# Saving Preprocessed data into a new csv file.
preprocessed.to_csv("training_data_preprocessed.csv", index=False)
print("Data saved Successfully!")
```
